```python
def neville(x_eval, x_datos, y_datos):# creamos la función del metodo neville, cuyas entradas son
    # el punto sobre el cual queremos saber su aproximación, los valores de x y los valores de f(x)
    n = len(x_datos)

    # Creamos una matriz Q de tamaño nxn llena de ceros
    Q = []
    for i in range(n):
        fila = [0.0] * n
        Q.append(fila)

    # Llenamos la primera columna de Q con los valores de f(x)
    for i in range(n):
        Q[i][0] = y_datos[i] #i es el numero de fila, y en la lista y_datos el subindice del valor de f(x)

    # Calculamos el resto de la tabla usando la formula del metodo. i el valor de la fila y j el valor de la columna de la matriz
    for i in range(1, n):
        for j in range(1, i + 1):
            numerador = (x_eval - x_datos[i - j]) * Q[i][j - 1] - (x_eval - x_datos[i]) * Q[i - 1][j - 1]
            denominador = x_datos[i] - x_datos[i - j]
            Q[i][j] = numerador / denominador

    return Q
```

```python
#  Datos de prueba  basados en la tabla del ejemplo del Burden
x_puntos = [1.0, 1.3, 1.6, 1.9, 2.2]
fx_puntos = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]
x_buscar = 1.5
```

```python
# Mandamos a llamar a la función con sus respectivos parametros
tabla_Q = neville(x_buscar, x_puntos, fx_puntos)
```

```python
#
for i in range(len(x_puntos)):
    # Imprimimos primero el valor de x_i
    print(f"{x_puntos[i]}\t", end="")# end="" esto hace que cada print se imprima a la derecha del anterior y no haya salto de linea
    #y \t da un espacio para el siguinte print
    for j in range(i + 1):
        print(f"{tabla_Q[i][j]}\t", end="")
    print() # Salto de línea para la siguiente fila
```

```python
n_ultimo = len(x_puntos) - 1
print(f"\nResultado final P({x_buscar}) = {tabla_Q[n_ultimo][n_ultimo]}")
# Y por ultimo imprimimos el ultimo valor de la diagonal, y definimos su subindice como n-1,n-1
#ya que en pyhon las columnas  y filas empiezan desde el subindice 0
```
