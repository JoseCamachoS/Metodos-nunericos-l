<font color="cyan">**Método de biseción**</font>

Queremos resolver la ecuacion $x^3+4x^2-10=0$ en el intervalo $[1,2]$
.Para aplicar el **metodo de bisección** verificamos que la función $f(x)$ evaluada en los extremos opuestos del intervalo tienen signos opuestos, y que la función sea continua en ese intervalo.Ambas cosas son ciertas, por lo tanto podemos usar el método.

<font color="red">Importamos las librerias necesarias</font>

```python
from math import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
  return x**3 +4*x**2-10.0 # se define la función de la que se aproximará su raiz
```

Probamos que la función sea correcta, y la fucnión evaluada en los extremos da signos opuestos, por lo tanto dentro del intervalo se encuentra una raiz por el teorema del valor intermedio



```python
f(1)
```

```python
f(2)
```

El siguiente fragmento de código en python genera la gráfica de la funcion para utilizar en el métdo de bisección

```python
x=np.linspace(0,3) #genera una particion
plt.plot(x,f(x))#muestra las parejas ordendas, (x,y)
plt.title('grafica f(x)')#titulo superior de la gráfica
plt.xlabel('x')
plt.ylabel('y')

plt.axhline(y=0,color='red')#trazar linea horizontal en Y=0
```

Ahora comenzamos con el algoritmo para implementar el método de bisección

```python
#ingreso de datos de entrada para los valores extremos del interbalo[a,b]
a=1
b=2

#guarda valores iniciales
a0=a
b0=b

tol=0.000001 #tolerancia del error
nmax=100 #numero maximo de iteraciones
i=0 #contador para mi numero de iteraciones
```

```python
# Método de Bisección

FA=f(a)  # se calcula f(a) y se guarda en la variable en FA

#código de la parte superior de la tabla
print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(i, a, f(a), b, f(b), a+(b-a)/2, f(a+(b-a)/2), (b-a)/2))

while i< nmax: #la condicion indica que el ciclo se repetirá mientras el contador sea menor que el numero maximo de iteraciones
  p=a+(b-a)/2 #se calcula el punto medio
  FP=f(p) # se calcula el valor de la funcion en el punto medio y se guarda en FP
  if FP==0 or (b-a)/2<tol:
    break   #si FP es 0, p es raiz y se termina el algoritmo, o si (b-a)/2 es menor que la torerancia se muestra p como raiz mas aproximada y se rompe el ciclo
  else:
    i+=1 #aumentamos uno al contador
    if FA*FP>0: #si se cumple la condicion, entre a y p no esta la raiz, pues no hay cabio de signp
      FA=FP     #asi que ahora hacemos a p nuestro nuevo a, y b queda igual
      a=p
    else:   # si no se cumple lo anterior entonces si hay cambios de signo entre f(a) y f(p), por lo tanto la raiz debe estar dentro de ese intervalo
      b=p   # y se toma ahora a p como el extremo b
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(i, a, FA, b, f(b), p, FP, (b-a)/2))
    #imprimimos valores de las variables para mostrar en la tabla

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,p))
# en caso de exceder el numero de iteraciones maximas se muestra la raiz p hasta la que se llegó
```
